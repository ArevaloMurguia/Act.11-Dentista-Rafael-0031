```python
# UIII Act 10
# IA con Lista de Diccionarios
# Tema: Dentista
# Rafael Arevalo
# Fecha: 25/May/26
# NL = 3

dentistas = []

# Funcion registrar
def registrar():
    dentista = {}

    print("\n--- REGISTRAR DENTISTA ---")

    dentista["nombre"] = input("Nombre: ")
    dentista["numero_colegiado"] = input("Numero de colegiado: ")
    dentista["clinica_propia"] = input("Clinica propia (si/no): ")
    dentista["servicios_ofrecidos"] = input("Servicios ofrecidos: ")
    dentista["materiales_stock"] = input("Materiales en stock: ")
    dentista["citas_dia"] = input("Citas del dia: ")
    dentista["anios_practica"] = input("Años de practica: ")

    dentistas.append(dentista)

    print("Dentista registrado correctamente")


# Funcion mostrar
def mostrar():

    if dentistas == []:
        print("\nNo hay dentistas registrados")

    else:
        print("\n================ LISTADO DE DENTISTAS ================")
        print(f"{'Nombre':15} {'Colegiado':12} {'Clinica':10} {'Citas':8} {'Años':8}")
        print("-" * 60)

        for d in dentistas:
            print(f"{d['nombre']:15} {d['numero_colegiado']:12} {d['clinica_propia']:10} {d['citas_dia']:8} {d['anios_practica']:8}")


# Funcion diagnostico IA
def diagnostico():

    print("\n--- DIAGNOSTICO IA ---")
    problema = input("Problema (caries/dolor/ortodoncia): ")

    if problema == "caries":
        print("IA: Aplicar resina")

    elif problema == "dolor":
        print("IA: Revisar nervio dental")

    elif problema == "ortodoncia":
        print("IA: Ajustar brackets")

    else:
        print("IA: Revisar con especialista")


# Menu principal
while True:

    print("\n=========== MENU ===========")
    print("1. Registrar dentista")
    print("2. Mostrar dentistas")
    print("3. Diagnostico IA")
    print("4. Salir")
    print("============================")

    op = input("Selecciona una opcion: ")

    if op == "1":
        registrar()

    elif op == "2":
        mostrar()

    elif op == "3":
        diagnostico()

    elif op == "4":
        print("Programa terminado")
        break

    else:
        print("Opcion incorrecta")
```
